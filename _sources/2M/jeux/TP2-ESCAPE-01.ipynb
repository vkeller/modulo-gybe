## Énigme 3 : Le Briseur de Code 

Voici un fichier nommé anti_virus.py. Il contient un algorithme de tri.

1. Copiez-le
2. Collez-le dans une fenêtre de l'éditeur Thonny
3. Enregistrez-le avec le nom : `anti_virus.py`
4. Exécutez-le

````python
# =====================================================================
# MISSION 3 : RECONSTRUIRE LE MANIFESTE ÉCOLOGIQUE
# =====================================================================
# Consignes :
# 1. Lancez le script et entrez le mot de passe trouvé à l'énigme 2.
# 2. Répondez à la question sur la complexité.
# 3. Corrigez l'erreur dans l'algorithme de tri pour décoder le message.
# =====================================================================

# --- ÉTAPE 1 : VÉRIFICATION DU MOT DE PASSE ---
mot_de_passe = input("Entrez le mot de passe secret de l'énigme 2 : ")

if mot_de_passe != "VERDURE":
    print("ACCÈS REFUSÉ. Le système s'est verrouillé.")
else:
    print("\n--- ACCÈS ACCÉPTÉ ---")
    print("Vérification de vos compétences en algorithmique...")
    
    # --- ÉTAPE 2 : QUESTION DE COMPLEXITÉ ---
    print("Ce programme utilise un tri à bulles (deux boucles imbriquées).")
    print("Quelle est sa complexité dans le pire des cas si la liste contient N éléments ?")
    print("Options disponibles : 'N', 'N^2', ou '1'")
    
    reponse = input("Votre réponse : ")
    
    if reponse != "N^2":
        print("Erreur d'analyse algorithmique. Fin du programme.")
    else:
        print("\n--- ANALYSE LOGIQUE CORRECTE ---")
        print("Début du processus de tri des données...")
        
        # Deux listes simples (les positions et les lettres correspondantes)
        # Le message d'origine est "ALERTE" mais il est complètement mélangé.
        positions = [5, 2, 6, 1, 4, 3]
        lettres = ["T", "L", "E", "A", "R", "E"]
        
        # Algorithme de tri à bulles pour remettre les positions dans l'ordre (1, 2, 3...)
        # ATTENTION : Il y a une erreur de logique ci-dessous ! Le tri se fait à l'envers.
        # Trouvez le symbole à changer pour trier du plus petit au plus grand.
        taille = 6
        for i in range(taille):
            for j in range(0, taille - i - 1):
                # L'erreur est sur la ligne juste en dessous :
                if positions[j] < positions[j+1]:
                    # On échange les positions
                    temporaire_pos = positions[j]
                    positions[j] = positions[j+1]
                    positions[j+1] = temporaire_pos
                    
                    # On échange AUSSI les lettres pour qu'elles suivent leur position
                    temporaire_lettre = lettres[j]
                    lettres[j] = lettres[j+1]
                    lettres[j+1] = temporaire_lettre
        
        # Reconstruction du message final sous forme de texte
        message_decode = ""
        for k in range(taille):
            message_decode = message_decode + lettres[k]
            
        print("\nRésultat du décodage :", message_decode)
````
